Word Embedding using One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
words = [['cat'],['dog'],['apple'],['car']]
```

```python
encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)
```

```python
print("words :",encoder.categories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [["python","is","a","language"],
 ["java","is","a","language"],
  ["python","is","easy"],
   ["java","is","easy"],
    ["python","is","powerful"],
     ["java","is","powerful"]]
```

```python
model = Word2Vec(sentences,vector_size=10,window=2,min_count=1,workers=1)
```

```python
print(model.wv.most_similar('java',topn=3))
```

```python
print(model.wv.most_similar('python',topn=4))
```

Task 3: Calculate Similarity Between Words

Use the trained Word2Vec model to calculate the similarity between two words.

```python
similarity = model.wv.similarity('python', 'java')
print("Similarity between python and java:", similarity)
```

Task 4: Display the Word Embedding Vector

Display the 10-dimensional Word2Vec embedding vector of the word 'python'.

```python
vector = model.wv['python']
print("Word embedding for 'python':")
print(vector)
print("Vector dimensions:", vector.shape)
```

Task 5: Find the Most Similar Word to a Given Word

Find the single word that is most similar to 'language' using the trained Word2Vec model.

```python
similar_word = model.wv.most_similar('language', topn=1)
print("Most similar word to 'language':", similar_word)
```
